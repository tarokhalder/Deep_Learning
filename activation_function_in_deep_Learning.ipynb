Import necessary Library

```python
import numpy as np
import matplotlib.pyplot as plt
```

**1. ReLU-Rectified Linear Unit(Hidden Layer)**

```python
def relu(x) :
  return np.maximum(0 , x)  # return maximum value between 0 and x
```

```python
relu(10)
```

```python
plt.scatter(np.arange(10) , relu(np.arange(10)))
plt.show()
```

```python
relu(-15)
```

```python
relu(np.array([0 , -5 , 10]))
```

**2.Leaky ReLU (Hidden Layer)**

---


it allows a small , positive gradient when the unit is not active

```python
def leaky_relu(x , alpha = 0.01):
  return np.where(x > 0 , x , x * alpha) # similar to ternary operator
```

```python
leaky_relu(15)
```

```python
plt.scatter(np.arange(10) , leaky_relu(np.arange(10)))
plt.show()
```

```python
leaky_relu(-15)
```

**3.ELU - Exponential Linear Unit(Hidden Layer)**
---
ELU has a small scope for negative values unlike ReLU to avoid the dying ReLU problem



```python
def elu(x , alpha = 0.01):
  return np.where(x > 0 , x , alpha * (np.exp(x) - 1))
```

```python
elu(5)
```

```python
elu(np.array([1 , 2 , -5]))
```

 **4.Sigmoid Function(Output Layer & Binarry classification)**

---
Sigmoid function is 1 / (1 + e^-x)


```python
def sigmoid(x):
  return 1 / (1 + np.exp(-x));
```

```python
sigmoid(500)
```

```python
sigmoid(np.array([1 , 5 , 6]))
```

**5.Softmax function(Output Layer & Multi class Problem)**


---
Softmax converts a real vector to a vector catagorical Probabilities


```python
def softmax(x):
    e_x = np.exp(x - np.max(x))
    return e_x / e_x.sum(axis = 0)
```

```python
tarok = softmax(np.array([2 , 1 , 3]))
sum(tarok)
```

```python
tarok
```

**6.Swish Layer(Hidden Layer)**

---
Swish is self-gated activation function



```python
def swish(x , beta = 0.9):
    return x * sigmoid(beta * x)
```

```python
swish(5)
```

```python
swish(np.array([1 , 5 , 6 , 7]))
```

**8.TanH(Hyperbolic tangent)**

```python
def tanh(x):
  return np.tanh(x);
```

```python
a = np.array([1 , 2 , 3 , 4 , 5 , 6])
output = tanh(a)
print(output)
```
